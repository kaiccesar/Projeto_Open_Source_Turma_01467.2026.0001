```python
import pandas as pd
import glob

emprestimos = glob.glob('Datasets/dados_emprestimos/*.csv')
```

```python
lista_emprestimos =[]

for arquivo in emprestimos:
    df_emprestimos = pd.read_csv(arquivo)
    lista_emprestimos.append(df_emprestimos)
    
df_emprestimo = pd.concat(lista_emprestimos, ignore_index=True)
```

```python
df_emprestimos['data_emprestimo'] = pd.to_datetime(df_emprestimos['data_emprestimo']).dt.strftime('d%/%m/%Y')
df_emprestimos['data_renovacao'] = pd.to_datetime(df_emprestimos['data_renovacao']).dt.strftime('d%/%m/%Y')
df_emprestimos['data_devolucao'] = pd.to_datetime(df_emprestimos['data_devolucao']).dt.strftime('d%/%m/%Y')

```

```python
df_emprestimos['matricula_ou_siape'] = df_emprestimos['matricula_ou_siape'].astype(str)
```

```python
df_emprestimos['ano'] = df_emprestimos['data_emprestimo'].str.split('/').str[2]
df_emprestimos['mes'] = df_emprestimos['data_emprestimo'].str.split('/').str[2]
```
